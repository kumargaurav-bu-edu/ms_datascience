```python
import matplotlib.pyplot as plt
from matplotlib.table import Table
import numpy as np

# Function to truncate text for display
def truncate_text(text, max_length=30):
    """Truncate text to max_length characters"""
    if pd.isna(text):
        return 'NaN'
    text_str = str(text)
    if len(text_str) > max_length:
        return text_str[:max_length-3] + '...'
    return text_str

# Function to save dataframe as PNG image
def save_df_as_image(df, filename, title="", figsize=(30, 8), fontsize=7, max_text_length=25):
    """
    Save a pandas DataFrame as a PNG image
    """
    # Create a copy and truncate long text values
    df_display = df.copy()
    for col in df_display.columns:
        if df_display[col].dtype == 'object':
            df_display[col] = df_display[col].apply(lambda x: truncate_text(x, max_text_length))
        else:
            # Format numeric values
            df_display[col] = df_display[col].apply(lambda x: f"{x:.2f}" if isinstance(x, (int, float)) and not pd.isna(x) else str(x))
    
    # Convert to string matrix for table
    cell_text = []
    for i in range(len(df_display)):
        row = [str(df_display.iloc[i, j]) for j in range(len(df_display.columns))]
        cell_text.append(row)
    
    fig, ax = plt.subplots(figsize=figsize)
    ax.axis('tight')
    ax.axis('off')
    
    # Create table
    table = ax.table(cellText=cell_text, 
                     colLabels=[col[:20] + '...' if len(col) > 20 else col for col in df_display.columns],
                     cellLoc='left',
                     loc='center',
                     bbox=[0, 0, 1, 1])
    
    table.auto_set_font_size(False)
    table.set_fontsize(fontsize)
    table.scale(1, 2.5)
    
    # Style the header row
    for i in range(len(df_display.columns)):
        table[(0, i)].set_facecolor('#2E7D32')
        table[(0, i)].set_text_props(weight='bold', color='white')
        table[(0, i)].set_height(0.08)
    
    # Alternate row colors for better readability
    for i in range(1, len(df_display) + 1):
        for j in range(len(df_display.columns)):
            if i % 2 == 0:
                table[(i, j)].set_facecolor('#f5f5f5')
            else:
                table[(i, j)].set_facecolor('white')
            table[(i, j)].set_height(0.06)
    
    if title:
        plt.suptitle(title, fontsize=18, fontweight='bold', y=0.995)
    
    plt.savefig(filename, dpi=300, bbox_inches='tight', facecolor='white', edgecolor='none')
    plt.close()
    print(f"✓ Saved: {filename}")

# Save first 10 rows as PNG
print("Saving first 10 rows as PNG...")
save_df_as_image(train_df.head(10), 'train_first_10_rows.png', 
                 title='First 10 Rows of Training Dataset', 
                 figsize=(32, 7), fontsize=6, max_text_length=20)

# Save statistics as PNG
print("\nSaving statistics as PNG...")
stats_df = train_df.describe()
save_df_as_image(stats_df, 'train_statistics.png', 
                 title='Dataset Statistics (Numerical Columns)', 
                 figsize=(20, 6), fontsize=8, max_text_length=15)

print("\n✅ Both images saved successfully!")
print("Files created:")
print("  - train_first_10_rows.png")
print("  - train_statistics.png")
```
