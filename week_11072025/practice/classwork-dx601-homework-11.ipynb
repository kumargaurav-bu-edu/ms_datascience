```python
import numpy as np

q20 = np.array([[2., 5., -3., 2.0],
                [-2., 1., 3., -2.],
                [4., 1., 0., 16.]], dtype=float)

def rref(A, tol=1e-12):
    A = A.copy().astype(float)
    rows, cols = A.shape
    r = 0
    for c in range(cols):
        # find pivot in column c, starting at row r
        piv = np.argmax(np.abs(A[r:rows, c])) + r
        if r >= rows or abs(A[piv, c]) <= tol:
            continue
        # swap
        if piv != r:
            A[[r, piv]] = A[[piv, r]]
        # normalize pivot row
        A[r] = A[r] / A[r, c]
        # eliminate all other rows
        for i in range(rows):
            if i != r:
                A[i] = A[i] - A[i, c] * A[r]
        r += 1
        if r == rows:
            break
    # clean tiny values to exact zeros
    A[np.abs(A) < tol] = 0.0
    return A

p20 = rref(q20)
print(p20)
```
