### Problem 15

Given the following data, set `p15` to the weighted variance of 

| Color | Shape | Score | Probability |
|---|---|---|---:|
| red | square | 3 | 0.250 |
| blue | circle | 4 | 0.125 |
| purple | line | 2 | 0.125 |
| purple | diamond | 5 | 0.25 |
| blue | triangle | 3 | 0.25 |

```python
# YOUR CHANGES HERE
scores = np.array([3, 4, 2, 5, 3])
weights = np.array([0.250, 0.125, 0.125, 0.25, 0.25])

# Weighted mean
weighted_mean = np.sum(scores * weights) / np.sum(weights)

deviations = scores - weighted_mean
squared_deviations = deviations ** 2
weighted_squared_deviations = squared_deviations * weights

p15 = np.sum(weighted_squared_deviations) / np.sum(weights)
```

### Problem 17

Compute the sample mean and variance of the 1990 vineyard yields.
Assuming that the yields follow a normal distribution with your computed parameters, what would the one-sided p-value of a yield of 13 lugs be?

Hint: use the [SciPy stats module](https://docs.scipy.org/doc/scipy/reference/stats.html) to calculate the p-values from the distribution.

```python
# YOUR CHANGES HERE
from scipy import stats
yields_1990 = vineyard["lugs_1990"]
sample_mean = yields_1990.mean()
sample_variance = yields_1990.var(ddof=1)
sample_std = np.sqrt(sample_variance)
normal_dist = stats.norm(loc=sample_mean, scale=sample_std)
p_value = 1 - normal_dist.cdf(13)
p17 = p_value
```

```python
p17
```
