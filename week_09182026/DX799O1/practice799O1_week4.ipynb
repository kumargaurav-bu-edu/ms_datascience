```python
import pandas as pd
import numpy as np
```

```python
arr = np.array([1, 2, 4])

# Calculate mean and standard deviation
mean = np.mean(arr)
std_dev = np.std(arr)

print(f"Mean: {mean:.2f}, Standard Deviation: {std_dev:.2f}")
```

```python
# standardize the array
standardized_arr = (arr - mean) / std_dev
print(
    "Standardized Array:",
    np.array2string(standardized_arr, formatter={"float_kind": lambda value: f"{value:.2f}"})
)
```

```python
# normalized version of the array
normalized_arr = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print(
    "Normalized Array:",
    np.array2string(normalized_arr, formatter={"float_kind": lambda value: f"{value:.2f}"})
)
```
