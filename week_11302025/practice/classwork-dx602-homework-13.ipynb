```python
import requests
from bs4 import BeautifulSoup
import urllib3

# Disable SSL warnings (optional, but recommended to suppress warning messages)
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
url = 'https://results.chicagomarathon.com/2025/?pid=list'

# Disable SSL verification to bypass certificate errors
response = requests.get(url, verify=False)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
rows = soup.find_all('li', attrs={'class': 'list-group-item'})

row = rows[2]

```

```python
# print(row)

values = [x.text for x in row.find_all('div')]
s = ''.join(values)
values = s.split()
print(values)

# print(s)



```
