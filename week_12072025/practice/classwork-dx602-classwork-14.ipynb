```python
## (A) if working with a local file on your disk: 
# open a file for reading:
# filename = 'time.txt' # Time
# # filename = 'romeo.txt' # Romeo and Juliet
# filename = 'grapes.txt' # Grapes of Wrath
# f = open(filename, 'r')
# read all text, split into list of words:
# text = f.read()

## (B) working with a remote file, download from WWW:
import requests 
import urllib3
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)  # Suppress SSL warning

url_time = "https://raw.githubusercontent.com/bu-cds-omds/dx602-examples/refs/heads/main/data/time.txt" # Time
# url_time = "https://raw.githubusercontent.com/bu-cds-omds/dx602-examples/refs/heads/main/data/romeo.txt"  # Romeo and Juliet
# url_time = "https://raw.githubusercontent.com/bu-cds-omds/dx602-examples/refs/heads/main/data/grapes.txt" # Grapes of Wrath
response = requests.get(url_time, verify=False)  # Disable SSL verification to handle certificate chain issues
text = response.text  # Use .text instead of .content to get string instead of bytes
```

```python
words = text.split()

words = [w.lower() for w in words]  # Call .lower() method (with parentheses) to convert to lowercase

print(f"found {len(words)} total words")
```

```python
distinct_words = set(words)
print("distinct words: ", len(distinct_words))
```

```python
for word in distinct_words:
    print(word)
```
