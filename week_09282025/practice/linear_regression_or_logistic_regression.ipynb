## 1. Linear Regression
Purpose:

* Predicts a continuous outcome.
* Models the relationship between one or more independent variables (predictors) and a continuous dependent variable (response).

How it works:

* Fits a straight line (or hyperplane for multiple variables) through the data points by minimizing the sum of squared errors.

Example:

* Predicting a person's weight given their height.
* Predicting house prices based on features like size, location, and number of rooms.

Mathematical form:
$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \ldots + \epsilon $$
where $y$ is continuous.

```python
# Linear Regression example
from sklearn.linear_model import LinearRegression
import numpy as np

# Example data: heights (in cm) and weights (in kg)
X = np.array([[150], [160], [170], [180], [190]])  # heights
y = np.array([50, 60, 65, 72, 80])  # weights

linear_model = LinearRegression()
linear_model.fit(X, y)

predicted_weight = linear_model.predict([[175]])
print(f"Predicted weight for 175 cm: {predicted_weight[0]:.2f} kg")
```

## 2. Logistic Regression
Purpose:

* Predicts a categorical outcome, usually binary (0 or 1, yes or no).
* Models the probability that a given input belongs to a particular category.

How it works:

* Uses the logistic function to map any real-valued number into the [0, 1] range.
* Outputs probabilities, which can be converted into class labels by applying a threshold (e.g., 0.5).

Example:

* Predicting whether an email is spam (yes/no).
* Predicting whether a patient has a disease (yes/no).

Mathematical form:
$$P(y=1|x) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + \ldots)}} $$

```python
# Logistic Regression example
from sklearn.linear_model import LogisticRegression
import numpy as np

# Example data: hours studied and pass/fail (1=pass, 0=fail)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

logistic_model = LogisticRegression()
logistic_model.fit(X, y)

prob_pass = logistic_model.predict_proba([[3.5]])[0][1]
print(f"Probability of passing with 3.5 hours studied: {prob_pass:.2f}")
```

### When to use which?

| Aspect                 | Linear Regression                  | Logistic Regression                  |
|------------------------|----------------------------------|------------------------------------|
| Outcome variable       | Continuous (e.g., price, weight) | Binary categorical (e.g., yes/no)  |
| Output                 | Real number (can be any value)   | Probability between 0 and 1         |
| Type of relation modeled| Linear                           | Log-odds (non-linear function)      |
| Example applications   | Price prediction, temperature forecasting | Spam detection, disease diagnosis |

## Key Terms
* Discriminant: a way of distinguishing between two classes.
* Linear discriminant function: a discriminant that is based on a linear equation.
* Support vector machine: a type of linear discriminant function that tries to account for the possibility of outliers. It also tries to maximize the distance between the discriminant line and the data samples.
