```python
# X is uniform between 0 and 1
# Y = 1 if X > 0.7, else Y = 0

# Probability that Y = 1
prob_y_equals_1 = 1 - 0.7  # P(X > 0.7) = 0.3

# Probability that Y = 0
prob_y_equals_0 = 0.7  # P(X <= 0.7) = 0.7

# Average value of Y = Expected Value
p10 = 1 * prob_y_equals_1 + 0 * prob_y_equals_0
# p10 = 1 * 0.3 + 0 * 0.7
# p10 = 0.3
```

```python
# If questions asked for average calculate exact value
# X is uniform on [2, 4]

# Case 1: X in [2, 3] → Y = 1
# Case 1: X in [2, 3]
# Y = 1 (always 1, regardless of X value)
# We DON'T need expected_x_case1 because Y doesn't depend on X!
expected_y_case1 = 1  # Y is always 1
# length of the subrange 3-2
# total range of X 4-2
prob_case1 = (3 - 2) / (4 - 2)  # 0.5 
expected_y_case1 = 1

# Case 2: X in (3, 4] → Y = X + 1
prob_case2 = (4 - 3) / (4 - 2)  # 0.5

# X in the range [3,4], since x is uniformely distributed
# min =3 , max = 4 , mean = (3+4) / 2
expected_x_case2 = (3 + 4) / 2  # 3.5
expected_y_case2 = expected_x_case2 + 1  # 4.5

# Total expectation
p18 = prob_case1 * expected_y_case1 + prob_case2 * expected_y_case2
# p18 = 0.5 * 1 + 0.5 * 4.5 = 2.75
```
