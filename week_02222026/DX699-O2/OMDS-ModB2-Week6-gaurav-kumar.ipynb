# 2. Weekly graph question

Suppose you wanted to show the nitrate and phosphate level in a water sample.  What are the advantages and disadvantages of showing this as an area plot, as opposed to two separate line graphs?  How would you adjust the graphs shown to improve the presentation?

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

### Analysis of Different Visualization Approaches:

**Chart 1 - Dual-axis line chart:**
- ✅ Shows both chemicals clearly with their own scales
- ✅ Easy to see individual trends
- ❌ Requires careful reading to understand scale differences
- **Best for:** Comparing trends when chemicals have different magnitudes

**Chart 2 - Same axis line chart:**
- ✅ Direct visual comparison of absolute levels
- ✅ Easy to see when one exceeds the other
- ✅ Simple and clean
- **Best for:** When both chemicals are on similar scales

**Chart 3 - Overlapping transparent areas:**
- ✅ Shows where both chemicals overlap (combined burden)
- ✅ Maintains visibility of both series
- ❌ Can still be somewhat visually busy
- **Best for:** Understanding cumulative impact while seeing individual trends

**Chart 4 - Multi-panel with ratio:**
- ✅ Shows individual trends AND their relationship
- ✅ The N:P ratio is scientifically meaningful for eutrophication
- ✅ Most informative for water quality assessment
- **Best for:** Complete understanding of water quality dynamics

### Recommendation:
For water quality analysis, **Chart 4 (multi-panel with ratio)** is most useful because:
1. The N:P ratio determines which nutrient limits algal growth
2. Different ratios favor different types of algae
3. Both individual levels and their relationship are important for management decisions

```python
np.random.seed(1)
time_series_2 =  100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series_2)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

```python
df = pd.DataFrame({'ser1': time_series, 'ser2': time_series_2}, index = datearray)
df.plot.area()
```

**Advantages of seperate line charts:**
* Each chemical's trend is cleary visible and easy to track over time
* Can directly compare whether nitrate is increasing or decreasing relative to phosphate
* Easier to read specific value for each chemical value at a give point in time

**Disadvatages**
* Harder to see combined total burden of both chemicals
* Difficult to visualize the relationship and ratio between the two chemicals
* Does not show how the presence of one chemical correlates with other

**Advantages of Area Plot:**
* Shows the cummulative effet
* Good for understading overall water quality

**Disadvantages**
* The stacked nature obscures the actual value of the top series
* Difficult to see if phosphorus is increasing or decreasing
* Cannot determine the ratio
* The visual overlap can be misleading

**Graph improvements**
1. Seperate line
    * Plot both on same axes with diffrent colors
    * Include clear legend and axis labels with units
    *  Consider adding a third panel showing N:P ration over time

2. Area charts
    * Use semi transparent colors
    * Add data labels at key points
    * Include refrence line for dangerous levels
