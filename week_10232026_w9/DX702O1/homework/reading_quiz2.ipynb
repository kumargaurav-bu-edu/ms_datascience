Question 1

Suppose we have two-way fixed effects with cities C1, C2, and C3 and individuals I1, I2, I3. If individual I3 has only lived in C3, and no one else has lived in C3, then the problem is that: 

**Option A** is correct: we cannot distinguish whether the effect comes from the coefficient of $I_3$ or the coefficient of $C_3$.

### Explanation

In a two-way fixed effects (TWFE) framework with individual effects ($\alpha_i$) and city effects ($\gamma_c$):

1. **Perfect Collinearity:** Because individual $I_3$ has only ever lived in city $C_3$, and no other individual has ever lived in $C_3$, the indicator variable for individual $I_3$ is identical to the indicator variable for city $C_3$ across all observations ($\mathbb{I}_{i = I_3} = \mathbb{I}_{c = C_3}$).
2. **Identification Failure:** Because these two regressors are perfectly collinear, the model cannot separately identify $\alpha_{I_3}$ and $\gamma_{C_3}$. It can only estimate their combined sum ($\alpha_{I_3} + \gamma_{C_3}$).

### Why the other options are incorrect

* **Option B:** TWFE does not require every individual to live in every city; it only requires a connected network of "movers" across locations to identify relative city effects.
* **Option C:** Having an equal number of individuals and cities is mathematically irrelevant to identification.
* **Option D:** A fundamental identification problem (perfect multicollinearity) exists for this pair of fixed effects.

Question 2

In random effects, 

**Option C** is correct: we assume that the beta coefficients come from the same random distribution.

### Explanation

In a random effects framework (such as random intercept or random coefficient models), individual- or group-specific effects are treated as random draws from a **common, shared probability distribution**—most typically a normal distribution with a shared mean and variance:

$$\beta_i \sim \mathcal{N}(\mu, \Sigma) \quad \text{or} \quad u_i \sim \text{i.i.d.} \, (0, \sigma_u^2)$$

This exchangeability assumption is what makes the model identifiable. By assuming the effects share an underlying distribution, the model pools information across units (often referred to as partial pooling or shrinkage) and estimates the overall hyperparameters (the shared mean and variance components).

### Why the other options are incorrect

* **Option A:** Random effects models are solvable and routinely estimated using Generalized Least Squares (GLS/FGLS), Maximum Likelihood Estimation (MLE), Restricted Maximum Likelihood (REML), or Bayesian methods.
* **Option B:** If every individual coefficient were drawn from an entirely distinct distribution with unique, unconstrained means and variances, the parameters could not be identified or pooled across units.
* **Option D:** Random coefficients are typically continuous and assumed to be normally distributed, not binary outcomes from a Bernoulli distribution.

Question 3


According to the text, when you have an interaction term, you would need _______ times as many observations to get enough statistical power. 

**Option D** is correct: **16**.

### Explanation

According to Nick Huntington-Klein’s *The Effect* (Chapter 13, citing Andrew Gelman):

> *"Even if the difference is as large as one group having an effect twice the size of the other group, you may need **sixteen** times as many observations to have adequate statistical power for the interaction term as you'd need for the main term."*

### The Intuition Behind the "Rule of 16"

1. **Standard Errors:** In a balanced $2 \times 2$ design, the standard error of a difference-in-differences (an interaction term) is twice ($2\times$) as large as the standard error of a main comparison.
2. **Effect Sizes:** Differences in effects between groups (interactions) are typically smaller than main effects—often assumed to be roughly half ($1/2$) the magnitude.
3. **Signal-to-Noise Ratio:** With half the effect size and double the standard error, the signal-to-noise ratio ($z$-score) for detecting the interaction is $4$ times smaller ($2 \div \frac{1}{2} = 4$).
4. **Sample Size Scaling:** Because standard errors shrink with the square root of sample size ($\sqrt{N}$), detecting an effect that is effectively $4$ times harder to identify requires squaring that factor:

$$4^2 = 16\text{ times as many observations}$$

Question 4

Which of the following would be a nonlinear regression:

Option A
Y = beta_0 + beta_2 * X^2 

Option B
Y = (beta_0 + beta_1 * X)^2

Option C
Y = beta_0 * log(X) 

Option D
(b) and (c) 

**Option B** is correct: $Y = (\beta_0 + \beta_1 X)^2$.

### Explanation

In statistics and econometrics, a model is classified as **linear** or **nonlinear** based on whether it is **linear in the parameters ($\beta$)**, not whether it is linear in the independent variables ($X$):

* **Linear in parameters:** The parameters appear only as first-order terms (multiplied by a variable or constant) and are added together. The partial derivative with respect to any parameter does not depend on unknown parameters.
* **Nonlinear in parameters:** The parameters appear raised to powers, inside functions, or multiplied by each other.

### Breakdown of the Options

* **Option B is nonlinear:** Expanding the expression gives:

$$Y = \beta_0^2 + 2\beta_0\beta_1 X + \beta_1^2 X^2$$



The parameters enter quadratically ($\beta_0^2, \beta_1^2$) and interact multiplicatively ($\beta_0\beta_1$). The partial derivatives ($\frac{\partial Y}{\partial \beta_0} = 2(\beta_0 + \beta_1 X)$) depend directly on the unknown parameters, requiring nonlinear estimation methods (such as Nonlinear Least Squares).
* **Option A is a linear regression:** Although $X$ is squared, the parameters $\beta_0$ and $\beta_2$ enter linearly. By defining a transformed variable $Z = X^2$, the model becomes $Y = \beta_0 + \beta_2 Z$, which is estimated directly via standard Ordinary Least Squares (OLS).
* **Option C is a linear regression:** Although $X$ undergoes a logarithmic transformation, the parameter $\beta_0$ enters strictly linearly. By setting $W = \log(X)$, the model becomes $Y = \beta_0 W$, a standard linear regression through the origin.
* **Option D is incorrect** because (c) is linear in its parameter.

Question 5


Heteroskedasticity means that:

Option A
The error term has a standard deviation that depends on the value of X. 

Option B
The error term has a standard deviation that does not depend on the value of X.

Option C
The error term has a mean that depends on the value of X. 

Option D
The error term has a mean that does not depend on the value of X. 

**Option A** is correct: the error term has a standard deviation that depends on the value of $X$.

### Explanation

**Heteroskedasticity** occurs when the variance (and consequently the standard deviation) of the unobserved error term $\varepsilon$ changes across different values of the independent variable $X$:

$$\operatorname{Var}(\varepsilon \mid X) = \sigma_i^2 \quad \text{or} \quad \operatorname{SD}(\varepsilon \mid X) = f(X)$$

In a plot of the residuals against $X$, heteroskedasticity typically appears as a "fan" or "cone" shape where the spread of the data points widens or narrows as $X$ increases.

### Why the other options are incorrect

* **Option B:** A standard deviation that remains constant regardless of the value of $X$ ($\operatorname{Var}(\varepsilon \mid X) = \sigma^2$) is the definition of **homoskedasticity**.
* **Option C:** If the conditional mean of the error term depends on $X$ ($E[\varepsilon \mid X] \neq 0$), it violates the zero conditional mean assumption, indicating **endogeneity** (e.g., omitted variable bias or functional form misspecification), not heteroskedasticity.
* **Option D:** An error term whose mean does not depend on $X$ ($E[\varepsilon \mid X] = 0$) satisfies the **exogeneity assumption** (strict/mean independence).
