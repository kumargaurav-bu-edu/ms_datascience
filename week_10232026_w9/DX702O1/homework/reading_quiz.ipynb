### Why Option D is Correct

* **The $k - 1$ Rule:** When modeling a categorical variable with $k$ distinct categories in a regression that includes an intercept term, exactly $k - 1$ dummy variables are needed. With two groups ($k = 2$), only $2 - 1 = 1$ indicator variable is required.
* **Arbitrary Reference Level:** Either group can serve as the baseline (reference) group:
* Using $X_A$ treats Group B as the baseline:

$$Y = \beta_0 + \beta_1 X_A + \epsilon$$



Group B has an expected value of $\beta_0$, and Group A has an expected value of $\beta_0 + \beta_1$.
* Using $X_B$ treats Group A as the baseline:

$$Y = \alpha_0 + \alpha_1 X_B + \epsilon$$



Group A has an expected value of $\alpha_0$, and Group B has an expected value of $\alpha_0 + \alpha_1$.


Both formulations are mathematically equivalent and yield identical fitted values, predictions, and residual sums of squares.
* **The Dummy Variable Trap:** Including both $X_A$ and $X_B$ in the presence of an intercept introduces **perfect multicollinearity**, since $X_A + X_B = 1$ for every observation. This causes the design matrix $(X^TX)$ to be non-invertible (singular), making it impossible to uniquely solve for the ordinary least squares (OLS) coefficients.
