````python
# Find left eigenvector with eigenvalue 1
eigenvals, eigenvecs = np.linalg.eig(q10.T)  # Transpose for left eigenvector
# Find index where eigenvalue ≈ 1
idx = np.argmax(np.isclose(eigenvals, 1))
# Get corresponding eigenvector
steady = eigenvecs[:, idx].real
# Normalize to sum to 1
p10 = steady / steady.sum()
````

## The systematic process - RREF
- The rule: always work top to bottom, left to right
You always start with the top-left entry and work down and to the right. The earlier example that started with row 3 was misleading. Here’s the correct process.


- Step 1: Find the leftmost non-zero column
- Step 2: Make the top entry in that column a pivot (1)
- Step 3: Eliminate entries below the pivot
- Step 4: Move to the next column and repeat
- Step 5: For RREF, eliminate entries above each pivot


## What is Gaussian Elimination?
Gaussian elimination converts a matrix to Row Echelon Form (REF) using row operations. It’s the forward phase of solving systems of equations.

Difference: Gaussian Elimination vs RREF
Gaussian Elimination → Row Echelon Form (REF)
Leading 1s, zeros below pivots
May have non-zero entries above pivots
Gauss-Jordan Elimination → Reduced Row Echelon Form (RREF)
REF + zeros above pivots

## What is Linear Dependence?
A row is linearly dependent if it can be written as a combination of other rows:
- Row 3 = coefficient1 × Row 1 + coefficient2 × Row 2
