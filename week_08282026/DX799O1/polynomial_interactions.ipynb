```python
# Import Required Libraries
# Import statsmodels for statistical modeling (Ordinary Least Squares regression)
import statsmodels.api as sm
# Import pandas for creating and manipulating DataFrames
import pandas as pd
# Import numpy for numerical operations and random number generation
import numpy as np
```

```python
# Create a DataFrame of 100 rows x 5 columns of random values drawn from a
# standard normal distribution (mean=0, std=1) — these are the 5 feature variables (X)
df_x = pd.DataFrame(np.random.normal(0,1,(100,5)))

# Create the target variable y: sum each row of X (axis=1 = across columns),
# which sets the true coefficient of every feature to 1.0, then add random
# Gaussian noise (mean=0, std=1) to simulate real-world measurement error
y_true = df_x.sum(axis=1) + np.random.normal(0,1,(100,))

# Add a column of 1s (named 'const') to the features so the regression
# can estimate an intercept term
df_x = sm.add_constant(df_x)

# Create an Ordinary Least Squares (OLS) linear regression model object
# with y_true as the dependent variable and df_x as the independent variables
model = sm.OLS(y_true, df_x)

# Fit the model to the data — computes the coefficient estimates that
# minimize the sum of squared residuals
model_fit = model.fit()

# Print the estimated coefficients — expect ~1.0 for each feature
# (matching the true relationship) plus a small intercept near 0
print(model_fit.params)
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
# Fix mixed column name types: df_x has int column names (0-4) plus the string
# 'const' added by sm.add_constant — sklearn requires all names to be one type,
# so convert them all to strings
df_x.columns = df_x.columns.astype(str)

# Create a LinearRegression model object (sklearn's OLS implementation)
model = LinearRegression()

# Fit the model: finds coefficients minimizing squared error between
# predictions and y_true. Note: sklearn fits its own intercept by default
# (fit_intercept=True), so the 'const' column is redundant here
model.fit(df_x, y_true)

# Display the learned coefficients — one per column of df_x
# (expect ~1.0 for each feature, matching the true relationship)
model.coef_  # This will give you the coefficients of the features
```

```python
# Create a DataFrame of 100 rows x 5 columns of random values drawn from a
# standard normal distribution (mean=0, std=1) — these are the 5 feature variables (X)
cols = ["A", "B", "C", "D", "E"]
df_x = pd.DataFrame(np.random.normal(0,1,(100,5)), columns=cols)

df_x_with_square = pd.concat((df_x, (df_x**2).rename(columns = {f"{var}": f"{var}2" for var in cols})), axis=1)
y_true = df_x_with_square.sum(axis=1) + np.random.normal(0,1,(100,))

df_x_with_square = sm.add_constant(df_x_with_square)

model = sm.OLS(y_true, df_x_with_square)
model_fit = model.fit()
print(model_fit.params)
```
