# Linear Regression Mathematics Breakdown

---

## Model Specification

**Regression Equation:**
$$\hat{y} = 0.32x + 5.00$$

* **Slope / Weight ($w$):** $0.32$ — The change in predicted $\hat{y}$ per unit increase in $x$.
* **Intercept / Bias ($b$):** $5.00$ — The expected baseline value of $\hat{y}$ when $x = 0$.
* **Sample Size ($n$):** $10$ data points.

---



## 1. Mean Squared Error (MSE)

Mean Squared Error calculates the average squared difference between actual target values ($y_i$) and predicted values ($\hat{y}_i$).

### Formula
$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \frac{SS_{res}}{n}$$

### Sample Data Table ($n = 10$)

| Point ($i$) | $x$ | $y$ (Actual) | $\hat{y}$ (Predicted) | Error ($y - \hat{y}$) | Squared Error $(y - \hat{y})^2$ |
| :---: | :---: | :---: | :---: | :---: | :---: |
| **1** | 1.0 | 3.50 | 5.32 | $-1.82$ | **3.31** |
| **2** | 2.0 | 7.06 | 5.64 | $+1.42$ | **2.00** |
| **3** | 3.0 | 4.85 | 5.96 | $-1.11$ | **1.24** |
| **4** | 4.0 | 8.61 | 6.28 | $+2.33$ | **5.41** |
| **5** | 5.0 | 5.69 | 6.60 | $-0.91$ | **0.83** |
| **6** | 6.0 | 7.53 | 6.92 | $+0.61$ | **0.37** |
| **7** | 7.0 | 5.02 | 7.24 | $-2.22$ | **4.95** |
| **8** | 8.0 | 9.48 | 7.56 | $+1.92$ | **3.69** |
| **9** | 9.0 | 7.17 | 7.88 | $-0.71$ | **0.50** |
| **10** | 10.0 | 8.71 | 8.20 | $+0.51$ | **0.26** |

### Calculation Steps

1. **Sum of Squared Residuals ($SS_{res}$):**
   $$SS_{res} = \sum_{i=1}^{10} (y_i - \hat{y}_i)^2$$
   $$SS_{res} = 3.31 + 2.00 + 1.24 + 5.41 + 0.83 + 0.37 + 4.95 + 3.69 + 0.50 + 0.26 = \mathbf{22.55}$$

2. **Compute MSE:**
   $$MSE = \frac{SS_{res}}{n} = \frac{22.55}{10} = \mathbf{2.255 \approx 2.26}$$

## 2. Total Sum of Squares ($SS_{tot}$)

$SS_{tot}$ measures the baseline variance in $y$ relative to its sample mean ($\bar{y}$), without using any model features ($x$).

### Calculation Steps

1. **Calculate Target Mean ($\bar{y}$):**
   $$\bar{y} = \frac{\sum_{i=1}^{10} y_i}{10} = \mathbf{6.76}$$

2. **Deviation Table ($y - \bar{y}$):**

---



## 3. Coefficient of Determination ($R^2$)

$R^2$ indicates the proportion of total variance in the dependent variable $y$ that is explained by the feature $x$.

### Formula
$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$$

### Calculation
$$R^2 = 1 - \frac{22.55}{52.55}$$
$$R^2 = 1 - 0.4291 = \mathbf{0.5709 \approx 0.57}$$



### Summary Interpretation
* **Total Variance ($SS_{tot} = 52.55$):** Unexplained error if we simply guess the average value $\bar{y}$ for all predictions.
* **Residual Variance ($SS_{res} = 22.55$):** Remaining unexplained error after fitting our model $\hat{y} = 0.32x + 5.00$.
* **Explained Variance ($SS_{reg} = 30.00$):** Reduction in variance achieved by using our regression line ($52.55 - 22.55 = 30.00$).
* **Model Accuracy ($R^2 = 57\%$):** The regression model successfully captures **$57\%$** of the total variance in the data.
