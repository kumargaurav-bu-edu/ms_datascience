```python
import math
import sys

import matplotlib.pyplot as plt
import numpy as np
```

```python
def p17(a, row_sums, column_sums):
    actual_row_sums = np.sum(a, axis=1)
    actual_col_sums = np.sum(a, axis=0)
    
    rows_match = np.all(actual_row_sums == row_sums)
    cols_match = np.all(actual_col_sums == column_sums)
    
    return rows_match and cols_match
```

```python
a = np.array([[0, 1, 2], 
              [3, 4, 5]])

row_sums = [3, 12]
column_sums = [3, 5, 7]

# Check rows:
# Row 0: 0 + 1 + 2 = 3 ✓ (matches row_sums[0])
# Row 1: 3 + 4 + 5 = 12 ✓ (matches row_sums[1])

# Check columns:
# Col 0: 0 + 3 = 3 ✓ (matches column_sums[0])
# Col 1: 1 + 4 = 5 ✓ (matches column_sums[1])
# Col 2: 2 + 5 = 7 ✓ (matches column_sums[2])

p17(a, row_sums, column_sums)  # True
```

```python
a = np.array([[0, 2, 4],  # Changed from [0, 1, 2]
              [3, 4, 5]])

row_sums = [3, 12]
column_sums = [3, 5, 7]

# Check rows:
# Row 0: 0 + 2 + 4 = 6 ✗ (expected 3)
# Row 1: 3 + 4 + 5 = 12 ✓

# Rows don't match, so return False

p17(a, row_sums, column_sums)  # False
```
