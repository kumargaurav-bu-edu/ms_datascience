```python
import pandas as pd
import numpy as np
from datetime import date, datetime, timedelta
import math
```

```python
df = pd.DataFrame([[1, 4], [2, 5], [3, 6], [4, 7]], columns = ['A', 'B'])
df.describe(percentiles = [1/3, 2/3])
```

```python
num_customers = 100
np.random.seed(0)
```

```python
df_bank = pd.DataFrame(columns = ["CustomerID"])
df_bank["CustomerID"] = [str(x) for x in np.arange(num_customers)]
df_bank
```

```python
start = datetime(1950, 1, 1)
end = datetime(2024, 1, 1)
numdays = (end - start).days
random_days = np.random.randint(0, numdays, size = num_customers)
```

```python
df_bank["BirthDate"] = start + pd.to_timedelta(random_days, unit='D')
df_bank
```

```python
# override the birthdate column to be a string format of the date
df_bank["BirthDate"] = df_bank["BirthDate"].dt.strftime('%Y-%m-%d')
df_bank
```

```python
def make_ssn_string(num):
    ssn_str = f'{num:09}'
    return ssn_str[0:3] + "-" + ssn_str[3:5] + "-" + ssn_str[5:9]
ssn_vector_func = np.vectorize(make_ssn_string)
df_bank["SSN"] = ssn_vector_func(np.random.randint(0, 999999999, size = num_customers))
df_bank
```

```python
df_bank["AccountID"] = np.random.randint(0, num_customers, size = num_customers)
df_bank
```

```python
# add a random number of days to the birthdate to get the account opened date lower=0 upper=80 years
random_days = np.random.randint(0, 365 * 80, size = num_customers)
df_bank["AccountOpened"] = (pd.to_datetime(df_bank["BirthDate"]) + pd.to_timedelta(random_days, unit='D')).dt.strftime('%Y-%m-%d')
df_bank
```

```python
df_bank.loc[0, "BirthDate"] = "1980"
df_bank.loc[1, "BirthDate"] = "no date"
df_bank.loc[2, "AccountID"] = np.nan
df_bank
```

```python
df_bank["AccountType"] = np.random.choice(["checking", "savings", "cd"], size = num_customers)
df_bank
```

```python
# Replaces row 99 (last row) of df_bank with row 0 (first row) from df
df_bank.loc[num_customers - 1] = df.loc[0]
df_bank
```

```python
df_bank.to_csv("bank_customers.csv", index=False)
```

```python
df_bank_loaded = pd.read_csv("bank_customers.csv")
```

```python
# The first few rows
df_bank_loaded.iloc[0:5]
```

```python
# Format check to filter valid dates
valid_dates = pd.to_datetime(df_bank_loaded["BirthDate"], format='%Y-%m-%d', errors='coerce')
df_bank_datefix = df_bank_loaded[valid_dates.notna()].copy()

# Convert AccountOpened to datetime as well
df_bank_datefix["AccountOpened"] = pd.to_datetime(df_bank_datefix["AccountOpened"], format='%Y-%m-%d', errors='coerce')
df_bank_datefix["BirthDate"] = pd.to_datetime(df_bank_datefix["BirthDate"], format='%Y-%m-%d')
df_bank_datefix
```

```python
# Find all duplicate rows and show their original index
df_bank_datefix_nonull = df_bank_datefix.dropna()
duplicate_rows = df_bank_datefix_nonull[df_bank_datefix_nonull.duplicated(keep=False)]
print(f"Number of duplicate rows: {len(duplicate_rows)}")
print(f"\nRow indices of duplicate rows:")
print(duplicate_rows.index.tolist())
df_bank_datefix_nonull_no_duplicates = df_bank_datefix_nonull.drop_duplicates()
# Display the full duplicate rows
duplicate_rows
```

```python
df_bank_datefix_nonull_no_duplicates
```

```python
# Check for duplicate AccountIDs
duplicate_mask = df_bank_datefix_nonull_no_duplicates['AccountID'].duplicated(keep=False)
duplicate_account_ids = df_bank_datefix_nonull_no_duplicates[duplicate_mask]

if len(duplicate_account_ids) > 0:
    print(f"Total rows with duplicate AccountIDs: {len(duplicate_account_ids)}")
    print(f"Unique AccountID values that are duplicated: {duplicate_account_ids['AccountID'].nunique()}")
    
    # Show counts for each duplicate AccountID
    print(f"\nCount of occurrences for each duplicate AccountID:")
    print(duplicate_account_ids['AccountID'].value_counts())
    
    # Show all rows with duplicate AccountIDs (sorted by AccountID for easier viewing)
    print(f"\nAll rows with duplicate AccountIDs:")
    duplicate_account_ids.sort_values('AccountID')
else:
    print("No duplicate AccountIDs found - all AccountIDs are unique!")
```

```python
# Remove duplicate AccountIDs, keeping the first occurrence of each AccountID
df_bank_datefix_nonull_no_duplicates_unique_accounts = df_bank_datefix_nonull_no_duplicates.drop_duplicates(subset=['AccountID'], keep='first')
df_bank_datefix_nonull_no_duplicates_unique_accounts
```

```python

# Create explicit copy to avoid warning
df_bank_datefix_nonull_no_duplicates_unique_accounts = df_bank_datefix_nonull_no_duplicates_unique_accounts.copy()

# Add Age column
df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'] = (pd.to_datetime(date.today()) - pd.to_datetime(df_bank_datefix_nonull_no_duplicates_unique_accounts['BirthDate'])).dt.days / 365.25

# Calculate statistics (rounded down)
mean_age = math.floor(df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'].mean())
median_age = math.floor(df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'].median())
mode_age = math.floor(df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'].mode()[0])

print(f"Mean customer age: {mean_age}")
print(f"Median customer age: {median_age}")
print(f"Mode customer age: {mode_age}")
```

```python
# Calculate IQR
Q1 = df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'].quantile(0.25)
Q3 = df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'].quantile(0.75)
IQR = Q3 - Q1

# Define outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1:.2f}")
print(f"Q3 (75th percentile): {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower bound (Q1 - 1.5*IQR): {lower_bound:.2f}")
print(f"Upper bound (Q3 + 1.5*IQR): {upper_bound:.2f}")

# Find outliers
outliers = df_bank_datefix_nonull_no_duplicates_unique_accounts[
    (df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'] < lower_bound) | 
    (df_bank_datefix_nonull_no_duplicates_unique_accounts['Age'] > upper_bound)
]

print(f"\nNumber of outliers: {len(outliers)}")

if len(outliers) > 0:
    print(f"\nOutlier details:")
    print(outliers[['Age', 'BirthDate', 'AccountType']].sort_values('Age'))
    
    # Count outliers by type
    print(f"\nOutliers by type:")
    print(f"  Very young (age < {lower_bound:.2f}): {len(outliers[outliers['Age'] < lower_bound])}")
    print(f"  Very old (age > {upper_bound:.2f}): {len(outliers[outliers['Age'] > upper_bound])}")
else:
    print("No outliers found!")
```

```python
# Step 1: One-hot encode AccountType
one_hot = pd.get_dummies(df_bank_datefix_nonull_no_duplicates_unique_accounts["AccountType"])

# Step 2: Join to original dataframe (replace df2 with your actual dataframe name)
df_bank_with_onehot = df_bank_datefix_nonull_no_duplicates_unique_accounts.join(one_hot)

# Step 3: Display to see the result
df_bank_with_onehot.iloc[0:5]

# Step 4: Convert to integers
df_bank_with_onehot[['checking', 'savings', 'cd']] = df_bank_with_onehot[['checking', 'savings', 'cd']].astype(int)

# Step 5: Verify
df_bank_with_onehot[['AccountType', 'checking', 'savings', 'cd']].head()
```

```python
#Create a copy for analysis
df_check = df_bank_with_onehot.copy()

#Create a new column for the age at which the account was opened
df_check['AgeAtAccountOpened'] = (df_check['AccountOpened'] - df_check['BirthDate']).dt.days / 365.25

# Check 1: Accounts opened in the future
future_accounts = df_check[df_check['AccountOpened'] > pd.to_datetime(date.today())]
print(f"Number of accounts opened in the future: {len(future_accounts)}")
if len(future_accounts) > 0:
    print(future_accounts[['BirthDate', 'AccountOpened', 'Age', 'AccountType']].head(10))


# Check 2: Accounts opened before birth
accounts_before_birth = df_check[df_check['AccountOpened'] <  df_check['BirthDate']]
print(f"\n2. Accounts opened before birth: {len(accounts_before_birth)}")
if len(accounts_before_birth) > 0:
    print(accounts_before_birth[['BirthDate', 'AccountOpened', 'Age', 'AccountType']].head(10))

#Check 3: Very young childrens (under 5) opening accounts
young_children = df_check[df_check['AgeAtAccountOpened'] < 5]
print(f"\n3. Very young children opening accounts: {len(young_children)}")
if len(young_children) > 0:
    print(young_children[['BirthDate', 'AccountOpened', 'Age', 'AccountType']].head(10))

#Check 4: Very old people (over 120) opening accounts
old_people = df_check[df_check['AgeAtAccountOpened'] > 120]
print(f"\n4. Very old people opening accounts: {len(old_people)}")
if len(old_people) > 0:
    print(old_people[['BirthDate', 'AccountOpened', 'Age', 'AccountType']].head(10))


# Check 5: Summary statistics
print(f"\n5. Summary Statistics:")
print(f"   Age when account opened - Min: {df_check['AgeAtAccountOpened'].min():.2f} years")
print(f"   Age when account opened - Max: {df_check['AgeAtAccountOpened'].max():.2f} years")
print(f"   Age when account opened - Mean: {df_check['AgeAtAccountOpened'].mean():.2f} years")


```
