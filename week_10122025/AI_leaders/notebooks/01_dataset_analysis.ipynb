# 01_dataset_analysis

Overview: load datasets from ./data, summary stats, missingness, basic visualizations for quality and insights.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from itertools import islice

sns.set(style='whitegrid')
sns.set_context('notebook', font_scale=1.0)
FIG_DIR = (Path.cwd() / 'figures')
FIG_DIR.mkdir(parents=True, exist_ok=True)
DEFAULT_DPI = 200

base = (Path.cwd().resolve().parent / 'data').resolve()
print('data base:', base)
print('figures dir:', FIG_DIR)

# Discover target files recursively in the five dataset subfolders
TARGET_FILES = {
    'bank_churners': 'BankChurners.csv',
    'credit_card_fraud_detection': 'creditcard.csv',
    'credit_score_classification': 'train.csv',
    'financial_distress_prediction': 'Financial Distress.csv',
    'insurance_claims_dataset': 'car_insurance_claim.csv',
}

def discover_files(base_dir: Path, targets: dict) -> dict:
    found = {}
    for folder, filename in targets.items():
        p = (base_dir / folder / filename).resolve()
        if p.exists():
            found[folder] = p
        else:
            print(f'WARN: Missing expected file for {folder}: {p}')
    return found

files = discover_files(base, TARGET_FILES)
print('datasets found:')
for k, v in files.items():
    print(' -', k, '->', v)

# Robust loader that handles common quirks
READ_KW = dict(low_memory=False)

def try_load(path: Path) -> pd.DataFrame | None:
    suf = path.suffix.lower()
    try:
        if suf in {'.csv', '.txt'}:
            # handle potential weird separators automatically
            try:
                return pd.read_csv(path, **READ_KW)
            except Exception:
                return pd.read_csv(path, sep=';', **READ_KW)
        if suf in {'.xlsx', '.xls'}:
            return pd.read_excel(path)
        if suf == '.parquet':
            return pd.read_parquet(path)
    except Exception as e:
        print('failed to load', path.name, '->', e)
    return None

# Utilities for data quality

def summarize_quality(df: pd.DataFrame) -> pd.DataFrame:
    n_rows = len(df)
    n_cols = df.shape[1]
    missing_perc = df.isna().mean() * 100
    missing_cols = (missing_perc > 0).sum()
    total_missing_perc = df.isna().values.mean() * 100
    duplicated_rows = int(df.duplicated().sum())
    numeric_cols = df.select_dtypes(include='number').shape[1]
    categorical_cols = df.select_dtypes(exclude='number').shape[1]
    return pd.DataFrame({
        'rows': [n_rows],
        'cols': [n_cols],
        'numeric_cols': [numeric_cols],
        'categorical_cols': [categorical_cols],
        'columns_with_missing': [missing_cols],
        'missing_overall_%': [round(total_missing_perc, 2)],
        'duplicate_rows': [duplicated_rows],
    })

def slugify(s: str) -> str:
    return ''.join(ch if ch.isalnum() or ch in '-_' else '_' for ch in s.lower())

def plot_missingness(df: pd.DataFrame, title: str, save_png: bool = True):
    name_slug = slugify(title)
    # Bar of % missing per column (top 15)
    na_pct = (df.isna().mean() * 100).sort_values(ascending=False)
    top = na_pct.head(15)
    fig, ax = plt.subplots(figsize=(7, 3), dpi=DEFAULT_DPI)
    sns.barplot(x=top.values, y=top.index, ax=ax, color='#4472c4')
    ax.set_xlabel('% missing')
    ax.set_ylabel('')
    ax.set_title(f'Missingness (top columns): {title}')
    fig.tight_layout()
    if save_png:
        out = FIG_DIR / f'{name_slug}_missing_bar.png'
        fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
        print('saved:', out)
    plt.close(fig)

    # Missingness matrix (sample up to 1,000 rows for speed)
    sample = df.sample(n=min(1000, len(df)), random_state=1) if len(df) > 0 else df
    fig, ax = plt.subplots(figsize=(6, 3), dpi=DEFAULT_DPI)
    sns.heatmap(sample.isna(), cbar=False, cmap='Greys', yticklabels=False, ax=ax)
    ax.set_title(f'Missingness matrix (sample): {title}')
    ax.set_xlabel('columns')
    ax.set_ylabel('')
    fig.tight_layout()
    if save_png:
        out = FIG_DIR / f'{name_slug}_missing_matrix.png'
        fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
        print('saved:', out)
    plt.close(fig)

def plot_distributions(df: pd.DataFrame, title: str, save_png: bool = True):
    name_slug = slugify(title)
    num = df.select_dtypes(include='number')
    if not num.empty:
        cols = list(num.columns[:4])
        rows = 2
        cols_n = 2
        fig, axes = plt.subplots(rows, cols_n, figsize=(8, 5), dpi=DEFAULT_DPI)
        axes = axes.flatten()
        for i, col in enumerate(cols):
            ax = axes[i]
            ax.hist(num[col].dropna(), bins=20, color='#4472c4', alpha=0.9)
            ax.set_title(col)
            ax.set_xlabel('')
            ax.set_ylabel('')
        # hide unused axes
        for j in range(len(cols), len(axes)):
            axes[j].axis('off')
        fig.suptitle(f'Numeric distributions: {title}', y=0.98)
        fig.tight_layout()
        if save_png:
            out = FIG_DIR / f'{name_slug}_numeric_dist.png'
            fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
            print('saved:', out)
        plt.close(fig)

    cat = df.select_dtypes(exclude='number')
    if not cat.empty:
        cols = list(cat.columns[:4])
        rows = 2
        cols_n = 2
        fig, axes = plt.subplots(rows, cols_n, figsize=(8, 6), dpi=DEFAULT_DPI)
        axes = axes.flatten()
        for i, col in enumerate(cols):
            ax = axes[i]
            vc = cat[col].astype('string').fillna('NA').value_counts().head(10)
            ax.barh(vc.index, vc.values, color='#70ad47')
            ax.invert_yaxis()
            ax.set_title(col)
            ax.set_xlabel('count')
            ax.set_ylabel('')
        for j in range(len(cols), len(axes)):
            axes[j].axis('off')
        fig.suptitle(f'Categorical distributions: {title}', y=0.98)
        fig.tight_layout()
        if save_png:
            out = FIG_DIR / f'{name_slug}_categorical_dist.png'
            fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
            print('saved:', out)
        plt.close(fig)

# Load all datasets and keep summaries
loaded = {}
summary_rows = []
for name, path in files.items():
    df = try_load(path)
    if df is None:
        print('SKIP: failed to load', name)
        continue
    loaded[name] = df
    print(f"\n=== {name} :: {path.name} ===")
    display(df.head())
    q = summarize_quality(df)
    q.insert(0, 'dataset', name)
    summary_rows.append(q)

# Build overall summary table
if summary_rows:
    overall_summary = pd.concat(summary_rows, ignore_index=True)
    display(overall_summary)

    # consolidated comparison chart (compact + save)
    fig, ax = plt.subplots(figsize=(8, 4), dpi=DEFAULT_DPI)
    melted = overall_summary.melt(id_vars='dataset', value_vars=['rows','cols','columns_with_missing','missing_overall_%','duplicate_rows'], var_name='metric', value_name='value')
    sns.barplot(data=melted, x='dataset', y='value', hue='metric', ax=ax)
    ax.set_title('Dataset comparison metrics')
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.tick_params(axis='x', rotation=15)
    fig.tight_layout()
    out = FIG_DIR / 'overall_comparison.png'
    fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
    print('saved:', out)
    plt.close(fig)

# Removed per-dataset charts per request
```

```python
# Dataset-level coverage and concise data quality visuals

from typing import Dict, Tuple

def compute_coverage_tables(loaded: Dict[str, pd.DataFrame]) -> Tuple[pd.DataFrame, Dict[str, pd.Series]]:
    rows = []
    per_dataset_coverage: Dict[str, pd.Series] = {}
    for name, df in loaded.items():
        cov = (df.notna().mean() * 100).astype(float).sort_values(ascending=False)
        per_dataset_coverage[name] = cov
        rows.append({
            'dataset': name,
            'cols': df.shape[1],
            'full_columns': int((cov == 100.0).sum()),
            'all_null_columns': int((cov == 0.0).sum()),
            'coverage_mean_%': round(float(cov.mean()), 2),
            'coverage_median_%': round(float(cov.median()), 2),
        })
    return pd.DataFrame(rows), per_dataset_coverage

coverage_overview, per_dataset_coverage = compute_coverage_tables(loaded)

# Merge in duplicate rate from overall_summary if available
if 'overall_summary' in globals():
    dup = overall_summary[['dataset','rows','duplicate_rows']].copy()
    coverage_overview = coverage_overview.merge(dup, on='dataset', how='left')
    coverage_overview['duplicate_rate_%'] = (coverage_overview['duplicate_rows'] / coverage_overview['rows'].replace(0, np.nan)) * 100
else:
    coverage_overview['rows'] = np.nan
    coverage_overview['duplicate_rows'] = np.nan
    coverage_overview['duplicate_rate_%'] = np.nan

# Compute full-columns share and DQI (0-100)
coverage_overview['full_columns_share_%'] = (coverage_overview['full_columns'] / coverage_overview['cols'].replace(0, np.nan)) * 100

w_cov, w_full, w_dup = 0.5, 0.3, 0.2
coverage_overview['data_quality_index'] = (
    w_cov * coverage_overview['coverage_mean_%'].fillna(0) +
    w_full * coverage_overview['full_columns_share_%'].fillna(0) +
    w_dup * (100 - coverage_overview['duplicate_rate_%'].fillna(0))
).clip(0, 100).round(2)

display(coverage_overview.sort_values('data_quality_index', ascending=False).reset_index(drop=True))

# Chart: columns per dataset (compact)
fig, ax = plt.subplots(figsize=(6, 3), dpi=DEFAULT_DPI)
sns.barplot(data=coverage_overview, x='dataset', y='cols', ax=ax, color='#4472c4')
ax.set_title('Columns per dataset')
ax.set_xlabel('')
ax.set_ylabel('columns')
ax.tick_params(axis='x', rotation=15)
fig.tight_layout()
out = FIG_DIR / 'columns_per_dataset.png'
fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
print('saved:', out)
plt.show(); plt.close(fig)

# Chart: overall missingness % per dataset (compact)
if 'overall_summary' in globals():
    fig, ax = plt.subplots(figsize=(6, 3), dpi=DEFAULT_DPI)
    sns.barplot(data=overall_summary, x='dataset', y='missing_overall_%', ax=ax, color='#c00000')
    ax.set_title('Overall missingness (%)')
    ax.set_xlabel('')
    ax.set_ylabel('% missing')
    ax.tick_params(axis='x', rotation=15)
    fig.tight_layout()
    out = FIG_DIR / 'overall_missingness.png'
    fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
    print('saved:', out)
    plt.show(); plt.close(fig)

# Chart: coverage buckets per dataset (compact, grouped bars)
BUCKETS = [
    ('100%', lambda s: (s == 100.0)),
    ('95-<100%', lambda s: (s >= 95.0) & (s < 100.0)),
    ('80-<95%', lambda s: (s >= 80.0) & (s < 95.0)),
    ('50-<80%', lambda s: (s >= 50.0) & (s < 80.0)),
    ('<50%', lambda s: (s < 50.0)),
]

bucket_rows = []
for name, cov in per_dataset_coverage.items():
    for label, pred in BUCKETS:
        count = int(pred(cov).sum())
        bucket_rows.append({'dataset': name, 'bucket': label, 'count': count})

bucket_df = pd.DataFrame(bucket_rows)
# ensure bucket order
bucket_order = [b[0] for b in BUCKETS]
fig, ax = plt.subplots(figsize=(8, 4), dpi=DEFAULT_DPI)
sns.barplot(data=bucket_df, x='dataset', y='count', hue='bucket', hue_order=bucket_order, ax=ax)
ax.set_title('Columns by coverage bucket')
ax.set_xlabel('')
ax.set_ylabel('columns')
ax.tick_params(axis='x', rotation=15)
fig.tight_layout()
out = FIG_DIR / 'coverage_buckets.png'
fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
print('saved:', out)
plt.show(); plt.close(fig)

# Chart: Data Quality Index per dataset
fig, ax = plt.subplots(figsize=(6, 3), dpi=DEFAULT_DPI)
order = coverage_overview.sort_values('data_quality_index', ascending=False)['dataset']
sns.barplot(data=coverage_overview, x='dataset', y='data_quality_index', order=order, ax=ax, color='#2f5597')
ax.set_title('Data Quality Index (0-100)')
ax.set_xlabel('')
ax.set_ylabel('DQI')
ax.tick_params(axis='x', rotation=15)
fig.tight_layout()
out = FIG_DIR / 'data_quality_index.png'
fig.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
print('saved:', out)
plt.show(); plt.close(fig)

# Tables: Top/Bottom coverage columns (5 each) per dataset (optional reference)
for name, cov in per_dataset_coverage.items():
    top5 = cov.head(5).round(2).rename('coverage_%').reset_index().rename(columns={'index': 'column'})
    bottom5 = cov.tail(5).round(2).rename('coverage_%').reset_index().rename(columns={'index': 'column'})
    print(f"\n{name}: top coverage columns")
    display(top5)
    print(f"{name}: bottom coverage columns")
    display(bottom5)
```

```python
# Render saved PNGs in a 3-per-row grid and export a combined overview
from PIL import Image
from math import ceil

# List of figure filenames in the order to display
png_files = [
    'overall_comparison.png',
    'columns_per_dataset.png',
    'overall_missingness.png',
    'coverage_buckets.png',
    'data_quality_index.png',
]

png_paths = [FIG_DIR / f for f in png_files if (FIG_DIR / f).exists()]

n = len(png_paths)
cols = 3
rows = ceil(n / cols)

fig, axes = plt.subplots(rows, cols, figsize=(12, 4*rows), dpi=DEFAULT_DPI)
axes = np.atleast_2d(axes)

for idx, path in enumerate(png_paths):
    r, c = divmod(idx, cols)
    ax = axes[r, c]
    img = Image.open(path)
    ax.imshow(img)
    ax.set_title(path.name)
    ax.axis('off')

# hide empty axes
for idx in range(n, rows*cols):
    r, c = divmod(idx, cols)
    axes[r, c].axis('off')

plt.tight_layout()
out = FIG_DIR / 'overview_grid.png'
plt.savefig(out, bbox_inches='tight', dpi=DEFAULT_DPI)
print('saved:', out)
plt.show()
```
