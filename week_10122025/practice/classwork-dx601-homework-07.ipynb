Great question! This is a common point of confusion. Let me clarify the difference between Problem 3 and Problem 4:

## Problem 3: Central Limit Theorem (Many identical variables)

**Situation**: 90 days of growth, each day is **the same type** of random variable
- Each day: mean = 1 cm, std = 0.5 cm
- You're summing **90 identical** random variables

**Formula for sum of n identical independent variables**:
```
If X₁, X₂, ..., Xₙ all have std = σ
Then: SD(X₁ + X₂ + ... + Xₙ) = σ × sqrt(n)
```

**Why?**
- Variance of each day = σ² = (0.5)²
- Total variance = n × σ² = 90 × (0.5)²
- Total std = sqrt(90 × σ²) = sqrt(90) × σ = σ × sqrt(90)

So: `p3 = 0.5 × sqrt(90)`

---

## Problem 4: Sum of Two Different Variables

**Situation**: Two growth periods that are **different**
- Period 1: mean = 1 lb, std = 0.5 lb
- Period 2: mean = 1.5 lb, std = 0.5 lb (different mean!)
- You're summing **2 different** random variables

**Formula for sum of different independent variables**:
```
SD(X + Y) = sqrt(Var(X) + Var(Y))
         = sqrt(SD(X)² + SD(Y)²)
```

**Why?**
- Variance of period 1 = (0.5)²
- Variance of period 2 = (0.5)²
- Total variance = (0.5)² + (0.5)²
- Total std = sqrt((0.5)² + (0.5)²)

So: `p4 = sqrt(0.5² + 0.5²)`

---

## Key Difference

| Problem | Variables | Formula |
|---------|-----------|---------|
| **Problem 3** | 90 **identical** variables | `σ × sqrt(n)` |
| **Problem 4** | 2 **different** variables | `sqrt(σ₁² + σ₂²)` |

**Problem 3**: You can factor out the common σ, so you get `σ × sqrt(n)`

**Problem 4**: The variables are different (different means), so you must add variances individually and then take the square root

Does this clarify the difference?
