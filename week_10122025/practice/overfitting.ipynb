```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Given data
features = [2, 4, 6, 8]
targets = [4, 8, 12, 18]

print("=" * 60)
print("🔍 OVERFITTING EXAMPLE ANALYSIS")
print("=" * 60)
print("Given data:")
for f, t in zip(features, targets):
    print(f"Feature: {f}, Target: {t}")

print("\n" + "=" * 60)
print("📊 ANALYZING THE THREE OPTIONS")
print("=" * 60)

# Option 1: target = 2 * feature
print("\n1️⃣ Option 1: Target = 2 × Feature")
print("Predictions:")
option1_predictions = []
for f, t in zip(features, targets):
    pred = 2 * f
    option1_predictions.append(pred)
    error = abs(t - pred)
    print(f"   Feature {f}: Predicted {pred}, Actual {t}, Error: {error}")

option1_total_error = sum(abs(t - p) for t, p in zip(targets, option1_predictions))
print(f"Total Error: {option1_total_error}")

# Option 2: target = 3 * feature  
print("\n2️⃣ Option 2: Target = 3 × Feature")
print("Predictions:")
option2_predictions = []
for f, t in zip(features, targets):
    pred = 3 * f
    option2_predictions.append(pred)
    error = abs(t - pred)
    print(f"   Feature {f}: Predicted {pred}, Actual {t}, Error: {error}")

option2_total_error = sum(abs(t - p) for t, p in zip(targets, option2_predictions))
print(f"Total Error: {option2_total_error}")

# Option 3: target = 2 * feature, except when feature = 8, then target = 3 * feature
print("\n3️⃣ Option 3: Target = 2 × Feature (except when Feature = 8, then Target = 3 × Feature)")
print("Predictions:")
option3_predictions = []
for f, t in zip(features, targets):
    if f == 8:
        pred = 3 * f  # Special case for feature = 8
    else:
        pred = 2 * f  # General rule
    option3_predictions.append(pred)
    error = abs(t - pred)
    print(f"   Feature {f}: Predicted {pred}, Actual {t}, Error: {error}")

option3_total_error = sum(abs(t - p) for t, p in zip(targets, option3_predictions))
print(f"Total Error: {option3_total_error}")

print("\n" + "=" * 60)
print("🎯 WHY OPTION 3 IS OVERFITTING")
print("=" * 60)

print("""
✅ Your answer is CORRECT! Here's why Option 3 represents overfitting:

🔍 OVERFITTING CHARACTERISTICS:
• Perfect fit on training data (Error = 0)
• Uses a complex rule with special exceptions
• Memorizes specific data points rather than learning general patterns
• Poor generalization to new data

📊 ANALYSIS:
• Option 1: Simple rule, some error (underfitting)
• Option 2: Simple rule, high error (underfitting)  
• Option 3: Complex rule with exception, perfect fit (OVERFITTING!)

🚨 THE PROBLEM WITH OPTION 3:
The model creates a special rule just for the point (8, 18).
This is memorization, not learning!

🔮 GENERALIZATION TEST:
What if we get a new data point with feature = 10?
• Option 1: Predicts 20 (consistent pattern)
• Option 2: Predicts 30 (consistent pattern)
• Option 3: Predicts 20 (but what if the true pattern has exceptions?)

The overfitted model (Option 3) might fail on new data because it
memorized the training data instead of learning the true pattern.
""")
```

```python
plt.figure(figsize=(15, 5))

# Plot 1: Option 1
plt.subplot(1, 3, 1)
plt.scatter(features, targets, color='red', s=100, label='Actual Data', zorder=5)
plt.plot(features, option1_predictions, 'b-', linewidth=2, label='Option 1: y = 2x')
for i, (f, t, p) in enumerate(zip(features, targets, option1_predictions)):
    plt.plot([f, f], [t, p], 'r--', alpha=0.7)
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title(f'Option 1: Simple Rule\nTotal Error: {option1_total_error}')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 2: Option 2
plt.subplot(1, 3, 2)
plt.scatter(features, targets, color='red', s=100, label='Actual Data', zorder=5)
plt.plot(features, option2_predictions, 'g-', linewidth=2, label='Option 2: y = 3x')
for i, (f, t, p) in enumerate(zip(features, targets, option2_predictions)):
    plt.plot([f, f], [t, p], 'r--', alpha=0.7)
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title(f'Option 2: Simple Rule\nTotal Error: {option2_total_error}')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 3: Option 3
plt.subplot(1, 3, 3)
plt.scatter(features, targets, color='red', s=100, label='Actual Data', zorder=5)
plt.plot(features[:3], option3_predictions[:3], 'purple', linewidth=2, label='y = 2x')
plt.plot([features[3]], [option3_predictions[3]], 'orange', marker='o', markersize=10, label='Special case: y = 3x')
plt.plot([features[3], features[3]], [2*features[3], option3_predictions[3]], 'orange', linestyle='--', linewidth=2)
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title(f'Option 3: Complex Rule (OVERFITTING)\nTotal Error: {option3_total_error}')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('/Users/gkumargaur/workspace/datascience/boston_university/practice/week10052025/overfitting_example.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n" + "=" * 60)
print("📈 VISUALIZATION SAVED")
print("=" * 60)
print("Graph saved as 'overfitting_example.png'")
print("The visualization shows why Option 3 overfits the data!")
```
