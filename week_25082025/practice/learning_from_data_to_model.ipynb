```python
import matplotlib.pyplot as plt
import numpy as np
import sklearn.linear_model

```

```python
X = [1,2,3,4,5]
X
```

```python
Y = [2, 4, 3, 4, 5]
Y
```

```python
plt.scatter(X, Y)
```

```python
m = sklearn.linear_model.LinearRegression()
m
```

```python
X = np.array(X).reshape(-1, 1)
X
```

```python
m.fit(X, Y)
```

```python
plt.scatter(X, Y, label="data")
plt.plot(X, m.predict(X), label="model")
plt.legend()
```
