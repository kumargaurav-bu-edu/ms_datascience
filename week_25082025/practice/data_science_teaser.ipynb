# Why Statistics?
A significant part of data science work is understanding data sets, and in particular their distribution. That is, what values or patterns are more common in the data set, and what values or patterns are less common or absent in the data set? Statistics are usually our first look at the distribution of the data set. We look at statistics of the data set with the belief that they are representative of the real world scenarios where the data set was collected.

# What is a Population?
An important notion in statistics is a population of interest. The population is not necessarily a set of people, but it is the set of all the people or objects or events that you are trying to understand. If you are a cardiologist, perhaps you are studying a population of overweight people considered at risk for heart disease. If you are an entomologist, perhaps it is all ants of a particular species. If you are a marine biologist, perhaps it is every 1 mL of water in the ocean. Whatever you study, there will be a population that you are seeking to understand.

# What is a Sample?
For most populations, it is not possible or feasible to get data for every member of the population. Instead, we collect data for a limited subset of the population that we call a sample of the population. We try to understand properties of the population by analyzing the data that we have from the sample.

# What is the Mean of a Data Set?
The mean or average of a data set is one way to measure a typical value in the data set. The mean of a data set does not necessarily match any particular value in the data set, but loosely speaking, the values of the data set will be around the mean.



```python
def mean(X):
    return sum(X) / len(X)

mean([3,4,8])




```

# What is the Standard Deviation of a Data Set?
The mean of a data set loosely describes a center of the data. The standard deviation of a data set describes whether the data is concentrated around that center.

```python
import math

def std(X):
    mean_local = mean(X)
    return math.sqrt(sum([(x - mean_local) ** 2 for x in X]) / (len(X) -1))

std([3,4,8])


```
