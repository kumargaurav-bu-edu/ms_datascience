```python
# In PCA, principal components are orthonormal (mutually orthogonal and unit length).
# Example: [1, 0] and [0, 1] are orthogonal since 1·0 + 0·1 = 0.
```

```python
abalone = pd.read_csv("../dx601-homework-08/abalone.tsv", sep="\t")
```

```python
from sklearn.decomposition import PCA

cols = abalone.select_dtypes(include=[np.number]).columns
X = abalone[cols].to_numpy()
X = (X - X.mean(axis=0)) / X.std(axis=0, ddof=0)

pc1 = PCA(n_components=1).fit(X).components_[0]
p16 = pd.DataFrame([pc1], columns=cols)
p16
```
