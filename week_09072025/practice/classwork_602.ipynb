```python
q9 = [{"color": "red", "score": 0.13},
      {"color": "blue", "score": 1.00},
      {"color": "green", "score": 0.64},
      {"color": "red", "score": 0.98},
      {"color": "blue", "score": 0.52},
      {"color": "violet", "score": 0.25}]
```

```python
colors = {}
[colors.setdefault(row["color"], []).append(row) for row in q9] #List comprehension , 
# each element is whatever the inner expression returns therefore its value will be None 
```

```python
# alternate way 
p9 = (lambda d={}: ([d.setdefault(r["color"], []).append(r) for r in q9], d)[1])()
print(p9)
```

```python
p9 = {}
for item in q9:
    color = item["color"]
    if color not in p9:
        p9[color] = []
    p9[color].append(item)

print(p9)
```

```python
from collections import defaultdict

p9 = defaultdict(list)
[p9[row["color"]].append(row) for row in q9]
p9 = dict(p9)  # convert back to a normal dict if you like

print(p9)
```

```python
q10 = {}

def add_data_10(row):
    if row['score'] < 0:
        return

    q10.setdefault(row['zipcode'], [])
    q10[row['zipcode']].append(row)
```

```python
row = {"zipcode": "red", "score": 1}
p10 = add_data_10(row)
print(p10)
```

```python
## Bisect example for sorting and inserting
from collections import namedtuple
from operator import attrgetter
from bisect import bisect, insort
from pprint import pprint

Movie = namedtuple('Movie', ('name', 'released','director'))
movies = [
    Movie('Jaws', 1975, 'Spielberg'),
    Movie('Titanic', 1997, 'Cameron'),
    Movie('The Birds', 1963, 'Hitchcock'),
    Movie('Aliens', 1986, 'Cameron')
]

# Find the first movie released after 1960
by_year = attrgetter('released')
movies.sort(key=by_year)
movies[bisect(movies, 1960, key=by_year)]
```

```python
# Insert a movie by maintaing a sort order
romance = Movie('Love Story', 1970, 'Hiller')
insort(movies, romance, key=by_year)
pprint(movies) 
```
