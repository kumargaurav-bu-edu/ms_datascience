```python
import math

def big_power_scientific(base: int, power: int, precision: int = 6) -> str:
    """
    Represent base**power in scientific notation (mantissa × 10^exponent)
    without computing the giant number directly.

    Args:
        base (int): the base number
        power (int): the exponent
        precision (int): digits after the decimal for mantissa

    Returns:
        str: representation in scientific notation
    """
    # log10(N) = power * log10(base)
    log10N = power * math.log10(base)

    # exponent (integer part of log10N)
    k = int(log10N)

    # mantissa = 10^(fractional part of log10N)
    fractional = log10N - k
    m = 10**fractional

    return f"{m:.{precision}f}e+{k}"

```

```python
print(big_power_scientific(6134123, 12312))
```
