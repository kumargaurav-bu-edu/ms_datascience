np.linalg.norm computes the length (norm) of vectors and matrices.
- For a 1-D array (vector): default is Euclidean (L2) norm.
- For a 2-D array (matrix): default is Frobenius norm (sqrt of sum of squares).
- You can change the norm with ord, compute along axes with axis, and keep dimensions with keepdims=True.

```python


v = np.array([3, 4])
np.linalg.norm(v)            # 5.0  (L2)

np.linalg.norm(v, 1)         # 7.0  (L1)
np.linalg.norm(v, np.inf)    # 4.0  (L∞)

A = np.array([[1, 2], [3, 4]])
np.linalg.norm(A)            # Frobenius norm
np.linalg.norm(A, 'fro')     # same as above
np.linalg.norm(A, 1)         # matrix 1-norm (max column sum)
np.linalg.norm(A, np.inf)    # matrix ∞-norm (max row sum)

# Norms per-row or per-column
np.linalg.norm(A, axis=1)    # row-wise vector L2 norms
np.linalg.norm(A, axis=0)    # column-wise vector L2 norms
```

```python
def unit_vector(v):
    v = np.asarray(v, dtype=np.float64)
    n = np.linalg.norm(v)
    print("n : ", n)
    if n == 0:
        raise ValueError("Cannot normalize zero vector")
    return v/n    

v_p10 = np.asarray([1.31, 2.13, -0.92])
p10 = unit_vector(v_p10)
print("p10 : ", p10)
```

## Orthogonal vectors
- Two vectors are orthogonal if their dot product is zero: u · v = 0 (angle is 90°).
- With floating point numbers, test “close to zero” with a tolerance.


```python
import numpy as np

def are_orthogonal(u, v, atol=1e-12):
    u = np.asarray(u, dtype=float)
    v = np.asarray(v, dtype=float)
    return np.isclose(u @ v, 0.0, atol=atol)

# example
are_orthogonal([1, 0, 0], [0, 3, 4])  # True
```

```python
base = np.array([1, 0, 0])
candidates = [np.array([1,1,1]), np.array([0.5,1,1]), np.array([0,1,1]), np.array([1,0,1])]
p11 = next(v for v in candidates if np.isclose(v @ base, 0.0))
print("p11 : ", p11)
```
